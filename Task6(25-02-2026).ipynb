```python
def smart_parking_system(capacity, vehicle_logs):
    """
    This function manages parking lot entries and exits.
    It calculates currently parked vehicles and checks capacity status.
    """
    
    parked = 0
    peak_usage = 0

    # Loop through vehicle logs
    for log in vehicle_logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1
        
        # Track peak usage
        if parked > peak_usage:
            peak_usage = parked

    # Check parking status
    if parked > capacity:
        status = "Over Capacity - Alert!"
    else:
        status = "Available"

    print("Currently Parked Vehicles:", parked)
    print("Peak Usage:", peak_usage)
    print("Parking Status:", status)

capacity = 50
vehicle_logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking_system(capacity, vehicle_logs)
```

```python
def estimate_delivery_time(distance, traffic, weather):
    """
    Estimates food delivery time based on distance,
    traffic conditions, and weather.
    """

    # Base time calculation (5 minutes per km)
    eta = distance * 5

    # Traffic adjustments
    if traffic.lower() == "high":
        eta += 15
    elif traffic.lower() == "medium":
        eta += 10
    elif traffic.lower() == "low":
        eta += 5

    # Weather adjustments
    if weather.lower() == "rainy":
        eta += 10
    elif weather.lower() == "stormy":
        eta += 20

    print("Estimated Delivery Time:", eta, "minutes")

estimate_delivery_time(8, "High", "Rainy")
```

```python
def theatre_occupancy(total_seats, booked_seats):
    """
    Calculates occupancy percentage and determines show status.
    """

    booked_count = len(booked_seats)
    occupancy_percentage = (booked_count / total_seats) * 100

    # Determine show status
    if occupancy_percentage == 100:
        status = "Housefull"
    elif occupancy_percentage >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    print("Occupancy:", round(occupancy_percentage), "%")
    print("Show Status:", status)

total_seats = 200
booked_seats = [1] * 150  # Simulating 150 booked seats

theatre_occupancy(total_seats, booked_seats)
```

```python
def classify_server_load(cpu_readings):
    """
    Classifies server load based on average CPU usage.
    """

    total = 0

    # Calculate total CPU usage
    for reading in cpu_readings:
        total += reading

    average = total / len(cpu_readings)

    # Classification rules
    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", round(average), "%")
    print("Server Status:", status)

cpu_readings = [45, 60, 70, 85, 90]

classify_server_load(cpu_readings)
```

```python
def classroom_resource_monitor(resource_usage):
    """
    Identifies overused classroom resources.
    """

    overused = []

    # Assume overuse if usage > 8 hours
    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

classroom_resource_monitor(resources)
```

```python
def event_registration_controller(capacity, registrations):
    """
    Controls event registration and prevents overbooking.
    """

    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

event_registration_controller(100, 105)
```
