```python
# function to check if recharge is valid or not
def recharge_system():
    
    # these are the only plans company supports
    valid_plans = [199, 299, 399, 599]
    
    # loop runs until user enters correct amount
    while True:
        
        amount = int(input("Enter recharge amount: "))
        
        # rule 1 → minimum recharge
        if amount < 50:
            print("Amount too low. Minimum recharge is 50.")
        
        # rule 2 → must match available plans
        elif amount not in valid_plans:
            print("Invalid plan selected.")
            print("Available plans are:", valid_plans)
        
        # if both conditions satisfied
        else:
            print("Recharge Successful ")
            break   # exit loop


# calling function
recharge_system()
```

```python
# function to check stock levels
def inventory_check(items):
    
    # going through each product one by one
    for product, qty in items.items():
        
        # if quantity is less than 15 → alert
        if qty < 15:
            print(product, ": Reorder needed!")
        
        # otherwise stock is fine
        else:
            print(product, ": Stock OK")


# sample dictionary (test data)
store = {
    "Milk": 10,
    "Bread": 25,
    "Eggs": 12,
    "Rice": 40
}

inventory_check(store)
```

```python
# function to calculate result based on average
def student_result(marks):
    
    total = 0
    
    # adding marks manually using loop
    for m in marks:
        total += m
    
    # calculating average
    avg = total / len(marks)
    print("Average =", avg)
    
    # checking pass or fail
    if avg >= 50:
        return "Pass"
    else:
        return "Fail"


# test list
marks_list = [65, 70, 40, 55]

print(student_result(marks_list))
```

```python
# function that calculates fare
def cab_fare(distance, peak_time):
    
    base_fare = 50      # fixed starting charge
    rate = 12           # price per km
    
    # basic fare calculation
    fare = base_fare + (distance * rate)
    
    # if peak hour → add 25% extra
    if peak_time.lower() == "yes":
        fare = fare + (fare * 0.25)
    
    return fare


# loop for retry option
while True:
    
    d = float(input("Enter distance travelled: "))
    p = input("Is it peak time? (yes/no): ")
    
    total = cab_fare(d, p)
    
    print("Total fare is ₹", round(total, 2))
    
    # asking if user wants to calculate again
    choice = input("Do you want to try again? ")
    
    if choice.lower() != "yes":
        print("Thank you for using fare calculator")
        break
```

```python
# function to check eligibility based on attendance
def check_attendance(record):
    
    present_days = 0
    
    # counting number of P's
    for day in record:
        if day == "P":
            present_days += 1
    
    # calculating percentage
    percentage = (present_days / len(record)) * 100
    
    print("Attendance % =", percentage)
    
    # eligibility rule
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# example input
data = ["P","P","A","P","P","P","A","P"]

print(check_attendance(data))
```

```python
# function to check password strength
def password_checker(pwd):
    
    special = "@#$"
    
    has_digit = False
    has_special = False
    
    # rule 1 → length
    if len(pwd) < 8:
        return "Password too short"
    
    # checking each character
    for ch in pwd:
        
        if ch.isdigit():
            has_digit = True
        
        if ch in special:
            has_special = True
    
    # final check
    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"


# user input
password = input("Enter password: ")
print(password_checker(password))
```
