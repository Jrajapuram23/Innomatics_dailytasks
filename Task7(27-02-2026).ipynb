```python
# Problem 1: Smart Light Controller
# This class represents a smart light that can be turned ON or OFF.

class SmartLight:
    
    # Constructor to initialize the light name and default status
    def __init__(self, light_name):
        self.light_name = light_name
        self.status = "OFF"  # Default state
    
    # Method to change light status
    def switch(self, action):
        if action.upper() == "ON":
            self.status = "ON"
        elif action.upper() == "OFF":
            self.status = "OFF"
        else:
            print("Invalid Action")
    
    # Method to display current status
    def display_status(self):
        print(f"{self.light_name} is {self.status}")


# Sample Usage
light = SmartLight("Bedroom Light")
light.switch("ON")
light.display_status()
```

```python
# Problem 2: Employee ID Card System
# This class stores employee details and prints an ID card format.

class Employee:
    
    # Constructor to store employee details
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    # Method to display ID card details
    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")


# Sample Usage
emp = Employee("Rahul", "EMP102", "IT")
emp.display_id_card()
```

```python
# Problem 3: Mobile Contact Record
# This class stores and displays contact details.

class Contact:
    
    # Constructor to initialize contact name and phone number
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    # Method to display saved contact information
    def display_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")


# Sample Usage
contact1 = Contact("Anita", "9876543210")
contact1.display_contact()
```

```python
# Problem 4: Product Price Tag Generator
# This class generates a formatted price tag for a product.

class Product:
    
    # Constructor to store product details
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price
    
    # Method to print price tag
    def print_price_tag(self):
        print(f"Product: {self.product_name}")
        print(f"Price: ₹{self.price}")


# Sample Usage
product1 = Product("Headphones", 2499)
product1.print_price_tag()
```

```python
# Problem 5: Movie Rating Display System
# This class stores movie details and displays rating information.

class Movie:
    
    # Constructor to initialize movie name and rating
    def __init__(self, movie_name, rating):
        self.movie_name = movie_name
        self.rating = rating
    
    # Method to display movie details
    def display_movie(self):
        print(f"Movie: {self.movie_name}")
        print(f"Rating: {self.rating} / 5")


# Sample Usage
movie1 = Movie("Inception", 4.8)
movie1.display_movie()
```

```python
# Problem 6: Delivery Address Manager
# This class manages customer delivery information.

class Delivery:
    
    # Constructor to store customer name and address
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    # Method to display delivery details
    def display_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")


# Sample Usage
delivery1 = Delivery("Suman", "Hyderabad")
delivery1.display_details()
```
