```python
# Logic Building Task – 1

# 1. User Login Check
# This function can be used to check if the input credentials match the actual ones.  

def login_check(input_username, input_password):
    username = "admin"
    password = "1234"

    if input_username == username and input_password == password: #if condition to give an output about credentials.
        print("Login Successful")
    else:
        print("Invalid Credentials")

login_check("admin", "1234")  #Calling the function with inputs. 
```

```python
# 2. Pass / Fail Analyzer
"""
 This function uses a for loop inside it and analyses:
 Total number of students
 Total number of students passed
 total number of students failed
 
 """

def pass_fail_analyzer(marks):
    pass_count = 0
    fail_count = 0

    for mark in marks:
        if mark >= 50:
            pass_count += 1
        else:
            fail_count += 1

    print("Total Students:", len(marks)) # prints total nuber of students.
    print("Passed:", pass_count) # prints number of students passes.
    print("Failed:", fail_count) # prints number of students failed.

marks = [45, 78, 90, 33, 60]
pass_fail_analyzer(marks)
```

```python
# 3. Simple Data cleaner
# This function cleans any given list of names by removing spaces and by converting to lower case


def clean_names(names):
    cleaned_list = []

    for name in names:
        cleaned_name = name.strip().lower() # removes spaces and converts names into lower case
        cleaned_list.append(cleaned_name) 

    print("Cleaned Names:", cleaned_list)

names = [" Alice ", "bob", " CHARLIE "]
clean_names(names)
```

```python
# 4. Message Length Analyzer
# This function can be used to analyse the length of any given message

def message_length_analyzer(messages):

    for message in messages:
        length = len(message)
        print("Message:", message) # prints the message
        print("Length:", length) # prints the length of the message

        if length > 10:
            print("This message is longer than 10 characters") # prints the error message if lenght is greater than 10


messages = ["Hi", "Welcome to the platform", "OK"]
message_length_analyzer(messages)
```

```python
#Error Message Detector
#This function can be used to count number of errors in a given log


def error_message_detector(logs):
    
    error_count = logs.count("ERROR")
    print("Total ERROR entries:", error_count)

logs = ["INFO", "ERROR", "WARNING", "ERROR"]
error_message_detector(logs)
```
