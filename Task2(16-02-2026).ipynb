```python
# Problem 1: Unique Words

# Take sentence input from user
sentence = input("Enter sentence: ")

# Split sentence into individual words using space as separator
words = sentence.split()

# Convert list to set to automatically remove duplicate words
unique_words = set(words)

# Display results
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)

```

```python
# Problem 2: Highest Salary

# Dictionary storing employee names as keys and salaries as values
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# max() with key=employees.get compares dictionary values instead of keys
highest_employee = max(employees, key=employees.get)

# Get salary of that employee
highest_salary = employees[highest_employee]

print("Highest Salary:", highest_employee, "-", highest_salary)

```

```python
# Problem 3: Max and Min

numbers = [45, 22, 89, 10, 66]

# Assume first element is both max and min initially
maximum = numbers[0]
minimum = numbers[0]

# Loop through list to compare values one by one
for num in numbers:
    if num > maximum:
        maximum = num   # update max if larger value found

    if num < minimum:
        minimum = num   # update min if smaller value found

print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)

```

```python
# Problem 4: Products above 1000

prices = [450, 1200, 899, 1500, 300]

count = 0  # counter to track how many prices exceed 1000

for price in prices:
    if price > 1000:
        count += 1  # increase counter when condition is satisfied

print("Products above 1000:", count)

```

```python
# Problem 5: Attendance Percentage

attendance = ["P", "P", "A", "P", "P"]

# Count how many days student was present
present_count = attendance.count("P")

# Calculate percentage using formula
percentage = (present_count / len(attendance)) * 100

print("Attendance Percentage:", percentage)

```

```python
# Problem 6: Unique Phone Numbers

phone_numbers = [9876543210, 9123456789, 9876543210]

# Convert list to set → removes duplicates automatically
unique_numbers = set(phone_numbers)

print("Unique phone numbers:", unique_numbers)

```

```python
# Problem 7: Character Frequency

text = "python"

frequency = {}  # empty dictionary to store character counts

for char in text:
    # If character already seen, increase count
    if char in frequency:
        frequency[char] += 1
    else:
        # Otherwise add it to dictionary with count 1
        frequency[char] = 1

print(frequency)
```

```python
# Problem 8: List to Tuple

numbers = [10, 20, 30]

# tuple() converts list into an immutable tuple
result = tuple(numbers)

print("Tuple:", result)

```

```python
# Problem 9: Check Key Exists

employee = {"name": "Ravi", "age": 25}

key = "name"

# 'in' checks whether key exists in dictionary
if key in employee:
    print("Employee exists")
else:
    print("Employee does not exist")

```

```python
# Problem 10: Average Marks

marks = [70, 80, 65, 75]

# sum() adds all marks, len() gives number of subjects
average = sum(marks) / len(marks)

print("Average Marks:", average)

```
