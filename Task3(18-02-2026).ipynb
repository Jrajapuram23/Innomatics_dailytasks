```python
# Dictionary containing employee names and their performance scores
employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# First we need to find the highest score in the dictionary
# values() gives all scores → max() returns highest
highest_score = max(employees.values())

# Now we check which employees have this highest score
top_performers = []

for name, score in employees.items():
    if score == highest_score:
        top_performers.append(name)

# Print result in required format
print("Top Performers Eligible for Bonus:", ", ".join(top_performers), f"(Score: {highest_score})")
```

```python
import string   # used for removing punctuation marks

query = "Buy mobile phone buy phone online"

# Convert entire sentence to lowercase so words match correctly
query = query.lower()

# Remove punctuation (like commas, dots, etc.)
query = query.translate(str.maketrans('', '', string.punctuation))

# Split sentence into individual words
words = query.split()

# Dictionary to store word frequency
word_count = {}

# Count each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

# Only keep words appearing more than once
result = {}

for word, count in word_count.items():
    if count > 1:
        result[word] = count

print(result)
```

```python
# List where index represents hour and value is sensor reading
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_readings = []

# enumerate() gives both index and value
for hour, value in enumerate(sensor_readings):

    # Even readings are considered valid
    if value % 2 == 0:
        # store as tuple (hour_index, reading_value)
        valid_readings.append((hour, value))

print("Valid Sensor Readings (Hour, Value):")
print(valid_readings)
```

```python
emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

# Loop through each email
for mail in emails:

    # Split email at '@' → domain is second part
    domain = mail.split("@")[1]

    # Count occurrences of each domain
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

# Total number of emails (needed for percentage calculation)
total_emails = len(emails)

# Display percentage usage of each domain
for domain, count in domain_count.items():
    percentage = (count / total_emails) * 100
    print(f"{domain}: {percentage:.0f}%")
```

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Average = sum of values / number of days
average_sales = sum(sales) / len(sales)

# A spike means 30% above average
threshold = average_sales * 1.3

# enumerate with start=1 because day numbering should start from 1
for day, amount in enumerate(sales, start=1):

    # If sale exceeds threshold → spike
    if amount > threshold:
        print(f"Day {day}: {amount}")
```

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

counts = {}

# Count how many times each user ID appears
for uid in user_ids:

    if uid in counts:
        counts[uid] += 1
    else:
        counts[uid] = 1

# Print only duplicate IDs (count > 1)
for uid, count in counts.items():
    if count > 1:
        print(f"{uid} → {count} times")
```
